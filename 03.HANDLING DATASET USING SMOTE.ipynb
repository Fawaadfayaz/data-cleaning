### SMOTE (Synthetic Minority Over-sampling Technique)

 HERE WE CREATE A SYNTHETIC / FAKE MORE MINORITY DATA , WE DID SOME SMALL VCAHNGES IN THAT MINORITY DATASET , NOT JUST DUPLICATING THE MINORTY DATASET 

```python
# we will import make_classification libraray here , to make our datset 

from sklearn.datasets import make_classification

x,y =make_classification (n_samples= 1000 ,n_redundant=0 , n_features= 2, n_clusters_per_class=1 , weights= [0.90] ,random_state= 12)
```

```python
import pandas as pd
df1 =pd.DataFrame (x, columns= ['f1', 'f2'])
df2 =pd.DataFrame (y, columns=['target'])
final_df = pd.concat([df1,df2], axis= 1)
final_df.head()
```

```python
# 4. Optional: See the imbalance visually
import matplotlib.pyplot as plt
plt.scatter(final_df['f1'], final_df['f2'], c=final_df['target'], cmap='coolwarm', alpha=0.7)
plt.title("Imbalanced Dataset (90% vs 10%)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
# now you can see that it is iumbalanced dataset now we will try to apply the SMOTE  to balance this
# #ist we will import the libraray as imblearen here 
!pip install imblearn
```

```python
from imblearn.over_sampling import SMOTE

oversample = SMOTE ()
x,y = oversample.fit_resample (final_df[['f1','f2']], final_df [['target']])
```

```python
x.shape #you will see 1800 datapoints it got oversampled 
```

```python
y.shape  #1800 datapoints also
```

```python
df1 =pd.DataFrame (x, columns= ['f1', 'f2'])
df2 =pd.DataFrame (y, columns=['target'])
oversampled_df = pd.concat([df1,df2], axis= 1)
```

```python
plt.scatter(oversampled_df['f1'], oversampled_df['f2'], c=oversampled_df['target'], cmap='coolwarm', alpha=0.7)
plt.title("Imbalanced Dataset (90% vs 10%)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

 here you can see that the minority data points got increased by just connecting
it just makes a line between two data points then plot the new synthetic datapoints on that line (or between that line)
and it now became a balanced dataset
